```python
import torch
from train_latent_transf_of_discrete_labels_fixed_sources import DDPM
from Network.masktransformer_fixed_clamp import MaskTransformer
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
mask_transf = MaskTransformer(img_size=28, hidden_dim=128, nclass=3, depth=7, heads=8, mlp_dim=1040, dropout=0.1, codebook_size=256, qk_norm=True, ignore_attn_to_source=True)
ddpm = DDPM(nn_model=mask_transf, betas=(1e-4, 0.02), n_T=500, device=device, drop_prob=0.1, n_classes=[2, 2, 2])
```

```python
ckpt_dir = '/work/dlclarge2/aliy-maskgit/concept_graphs/output/single-body_2d_3classes/DiT/latent/discrete/H32-train1/18-03-18-10_5000_1.6_256_500_2000_0.0001_None_1500_2.0_1/checkpoints'
ckpt_file = "ddpm_epoch_600.pth"
ckpt = torch.load(ckpt_dir + '/' + ckpt_file, map_location=device)
ddpm.load_state_dict(ckpt['model_state_dict'])
ddpm.to(device)
ddpm.eval()
```

```python
for name, param in ddpm.named_parameters():
    
    if param.requires_grad:
        print(f"Parameter: {name} | Shape: {param.shape} | Mean: {param.mean().item()} | Std: {param.std().item()}")

    if torch.isnan(param).any():
        print(f"NaN detected in parameter: {name}")
```
